```python
import json
import re
import tensorflow as tf
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
model_path = '/'

gru_model = load_model(model_path + 'gru_word2vec_model.keras')

bilstm_model = load_model(model_path + 'bilstm_word2vec_model.keras')

print("The models loaded sucessfully")
```

```python
with open(model_path + 'word_to_index.json' , 'r') as f :
  word_to_index = json.load(f)
max_len = 21
print("Vocabulary loaded" , len(word_to_index))
```

```python
def prepare_input(sentence) :
  sentence = sentence.lower()
  sentence = re.sub(r"[^\w\s]" , "" , sentence)
  words = sentence.split()
  sequence = []
  for word in words :
    if word in word_to_index :
      sequence.append(word_to_index[word])
  padded = pad_sequences([sequence] , maxlen = max_len , padding = 'post' , truncating = 'post')
  return padded
```

```python
def predict_both (sentence) :
  padded = prepare_input(sentence)
  gru_probability = gru_model.predict(padded , verbose = 0)[0][0]
  bilstm_probability = bilstm_model.predict(padded , verbose = 0 )[0][0]
  def get_intent(probability) :
    if(probability >= 0.5) :
      return "Non-Navigation"
    else :
      return "Navigation"
  print("Input Sentence :- " , sentence)
  # print("GRU Probability :- " , gru_probability)
  # print("BILSTM Probability :- " , bilstm_probability)
  # print("GRU Probability:", f"{gru_probability * 100:.2f}%")
  # print("BiLSTM Probability:", f"{bilstm_probability * 100:.2f}%")
  print("GRU Intent :- " , get_intent(gru_probability))
  print("BILSTM Intent :- " , get_intent(bilstm_probability))
  print("----------------------------------------------------------------------")
```

```python
custom_sentence = input("Enter the sentence whose intent(Naviagtion / Non-navigation) you want to know :- ")
predict_both(custom_sentence)
```
